```python
# ============================================================================
# 09_significance_tests.ipynb
# ----------------------------------------------------------------------------
# Beyond Full-Schema Prompting: A Graph-based Semantic Layer for Text-to-SQL
#
# Confirmatory statistics for the paper's central claims:
#   (1) EX PARITY: C4 (Hybrid) is statistically indistinguishable from C1 (Full)
#       on EX, per dataset x model -> McNemar paired test on per-question EX,
#       plus a TOST equivalence test (are they equivalent within a margin?).
#   (2) TOKEN SAVINGS: C4 (and C5_L2) use significantly fewer context tokens than
#       C1 -> paired test on per-question context tokens + bootstrap CI on % saved.
#   (3) PRUNING EFFECT: pruning strength lowers EX (main effect) while the
#       strength x model interaction is NOT significant -> OLS recap + bootstrap.
#   (4) BOOTSTRAP CIs for headline EX numbers.
#
# Consumes : results/tables/generation_results.csv  (notebook 02, full run)
# Produces : 09_parity_mcnemar.{csv,tex}
#            09_equivalence_tost.{csv,tex}
#            09_token_savings_tests.{csv,tex}
#            09_bootstrap_ex_ci.{csv,tex}
#            09_ex_ci_forest.{png,pdf}          forest plot of EX with 95% CI
#
# Anonymity: no filesystem paths are printed.
# ============================================================================


# %% [markdown]
# ## Cell 1 - Load configuration + results

# %%
%run 00_setup_and_config.ipynb


# %%
from scipy import stats

def load_results():
    full = TABLES_DIR / "generation_results.csv"
    if full.exists():
        df = pd.read_csv(full); print(f"Loaded full results: n={len(df)}.")
    else:
        raise FileNotFoundError("Run notebook 02 (full) first.")
    return df

res = load_results()
print("Datasets:", sorted(res["dataset"].unique()),
      "| Conditions:", sorted(res["condition"].unique()),
      "| Models:", sorted(res["model_tier"].unique()))


# %% [markdown]
# ## Cell 2 - Parameters
# EQUIV_MARGIN: the EX difference (in accuracy points) below which we treat two
# conditions as practically equivalent. 0.02 = 2 percentage points is a common,
# defensible choice; adjust to your field norm and justify in the paper.

# %%
EQUIV_MARGIN = 0.02          # equivalence margin for TOST (in EX units)
N_BOOT = 5000                # bootstrap resamples
ALPHA = 0.05
BASELINE = "C1"
PROPOSED = "C4"
np.random.seed(GLOBAL_SEED)


# %% [markdown]
# ## Cell 3 - Helper: aligned per-question EX vectors for two conditions
# Returns paired 0/1 EX arrays for the same qids under two conditions, for a
# given dataset x model. Only EX-scoreable rows (non-NaN) that appear in BOTH
# conditions are kept, so the test is properly paired.

# %%
def paired_ex(df, dataset, model_tier, cond_a, cond_b):
    sub = df[(df["dataset"] == dataset) & (df["model_tier"] == model_tier)]
    piv = sub.pivot_table(index="qid", columns="condition",
                          values="exec_match", aggfunc="first")
    if not {cond_a, cond_b}.issubset(piv.columns):
        return None, None
    pair = piv[[cond_a, cond_b]].dropna()
    if len(pair) == 0:
        return None, None
    return pair[cond_a].astype(int).values, pair[cond_b].astype(int).values


# %% [markdown]
# ## Cell 4 - (1) EX parity: McNemar paired test (C4 vs C1)
# McNemar focuses on discordant pairs (one condition right, the other wrong).
# H0: the two conditions have equal error rates. A NON-significant p (> alpha)
# is consistent with parity (but see the TOST in Cell 5 for a positive claim).

# %%
def mcnemar_exact(b, c):
    """Exact binomial McNemar on discordant counts b, c. Returns p (two-sided)."""
    n = b + c
    if n == 0:
        return 1.0
    return stats.binomtest(min(b, c), n, 0.5, alternative="two-sided").pvalue

rows = []
for ds in sorted(res["dataset"].unique()):
    for mt in MODEL_TIER_ORDER:
        a, b_ = paired_ex(res, ds, mt, BASELINE, PROPOSED)
        if a is None:
            continue
        c1_only = int(((a == 1) & (b_ == 0)).sum())   # C1 right, C4 wrong
        c4_only = int(((a == 0) & (b_ == 1)).sum())   # C4 right, C1 wrong
        p = mcnemar_exact(c1_only, c4_only)
        rows.append({
            "dataset": ds, "model_tier": mt, "n_pairs": len(a),
            "EX_C1": round(a.mean(), 4), "EX_C4": round(b_.mean(), 4),
            "delta_EX": round(b_.mean() - a.mean(), 4),
            "C1_only_correct": c1_only, "C4_only_correct": c4_only,
            "mcnemar_p": round(p, 4),
            "verdict": "parity (n.s.)" if p > ALPHA else "differs",
        })
parity = pd.DataFrame(rows)
display(parity)
save_table(parity, "09_parity_mcnemar")


# %% [markdown]
# ## Cell 5 - (1b) Equivalence test (TOST) for C4 vs C1
# A non-significant McNemar does NOT by itself prove equivalence. TOST (two one-
# sided tests) positively tests H1: |EX_C4 - EX_C1| < EQUIV_MARGIN. We run TOST on
# the paired per-question differences (d = EX_C4 - EX_C1) using a paired-t /
# normal approximation. p_tost < alpha => statistically equivalent within margin.

# %%
def tost_paired(a, b, margin):
    """
    TOST on paired differences d=b-a. Returns (mean_diff, p_tost, equivalent).
    Uses the paired-t framework: equivalence holds if both one-sided tests reject.
    """
    d = b.astype(float) - a.astype(float)
    n = len(d)
    md = d.mean()
    sd = d.std(ddof=1)
    if sd == 0:
        # zero variance: equivalent iff mean diff strictly inside margin
        eq = abs(md) < margin
        return md, (0.0 if eq else 1.0), eq
    se = sd / np.sqrt(n)
    # H0_lower: d <= -margin   ; H0_upper: d >= +margin
    t_lower = (md + margin) / se     # test that d > -margin
    t_upper = (md - margin) / se     # test that d < +margin
    p_lower = 1 - stats.t.cdf(t_lower, df=n - 1)   # P(T > t_lower)
    p_upper = stats.t.cdf(t_upper, df=n - 1)       # P(T < t_upper)
    p_tost = max(p_lower, p_upper)
    return md, p_tost, (p_tost < ALPHA)

rows = []
for ds in sorted(res["dataset"].unique()):
    for mt in MODEL_TIER_ORDER:
        a, b_ = paired_ex(res, ds, mt, BASELINE, PROPOSED)
        if a is None:
            continue
        md, p_tost, eq = tost_paired(a, b_, EQUIV_MARGIN)
        rows.append({
            "dataset": ds, "model_tier": mt, "n_pairs": len(a),
            "mean_diff_C4_minus_C1": round(md, 4),
            "equiv_margin": EQUIV_MARGIN,
            "p_tost": round(p_tost, 4),
            "equivalent": "yes" if eq else "no",
        })
tost = pd.DataFrame(rows)
display(tost)
save_table(tost, "09_equivalence_tost")
print(f"TOST margin = +/-{EQUIV_MARGIN} EX. 'equivalent=yes' means C4 is "
      f"statistically within the margin of C1.")


# %% [markdown]
# ## Cell 6 - (2) Token savings: paired test + bootstrap CI on % saved
# Context tokens are (nearly) deterministic per (qid, condition), so we compare
# per-question context_tokens_est between C1 and each economical condition. We use
# a Wilcoxon signed-rank test (non-parametric, robust) and bootstrap the mean %
# saved for a CI. Reported per dataset (tokens are model-independent).

# %%
def token_savings(df, dataset, cond, baseline=BASELINE):
    sub = df[df["dataset"] == dataset]
    piv = sub.pivot_table(index="qid", columns="condition",
                          values="context_tokens_est", aggfunc="first")
    if not {baseline, cond}.issubset(piv.columns):
        return None
    pair = piv[[baseline, cond]].dropna()
    base_tok = pair[baseline].values.astype(float)
    cond_tok = pair[cond].values.astype(float)
    # Wilcoxon signed-rank (paired)
    try:
        w_p = stats.wilcoxon(base_tok, cond_tok).pvalue
    except Exception:
        w_p = np.nan
    # per-question % saved, bootstrap mean CI
    pct = 100.0 * (base_tok - cond_tok) / base_tok
    boot = [np.mean(np.random.choice(pct, size=len(pct), replace=True))
            for _ in range(N_BOOT)]
    lo, hi = np.percentile(boot, [2.5, 97.5])
    return {
        "dataset": dataset, "condition": cond, "n": len(pair),
        "mean_tokens_C1": round(base_tok.mean(), 1),
        "mean_tokens_cond": round(cond_tok.mean(), 1),
        "mean_pct_saved": round(pct.mean(), 2),
        "pct_saved_CI_low": round(lo, 2), "pct_saved_CI_high": round(hi, 2),
        "wilcoxon_p": (f"{w_p:.2e}" if not np.isnan(w_p) else "n/a"),
    }

rows = []
for ds in sorted(res["dataset"].unique()):
    for cond in ["C4", "C5_L1", "C5_L2"]:
        r = token_savings(res, ds, cond)
        if r:
            rows.append(r)
tok_tests = pd.DataFrame(rows)
display(tok_tests)
save_table(tok_tests, "09_token_savings_tests")


# %% [markdown]
# ## Cell 7 - (3) Pruning strength effect + interaction recap (OLS)
# Recap the OLS with strength (0..3 over C4->C5_L3), model rank, and interaction,
# per dataset. Reports the strength main effect (expected negative, significant)
# and the strength x model interaction (expected non-significant).

# %%
STRENGTH_ORDER = ["C4", "C5_L1", "C5_L2", "C5_L3"]
STRENGTH_INDEX = {c: i for i, c in enumerate(STRENGTH_ORDER)}
TIER_RANK = {"WEAK": 0, "MID": 1, "STRONG": 2}

def ols_interaction(df, dataset):
    d = df[(df["dataset"] == dataset) &
           (df["condition"].isin(STRENGTH_ORDER))].dropna(subset=["exec_match"]).copy()
    if not len(d):
        return None
    d["strength"] = d["condition"].map(STRENGTH_INDEX)
    d["model_rank"] = d["model_tier"].map(TIER_RANK)
    d["inter"] = d["strength"] * d["model_rank"]
    X = np.column_stack([np.ones(len(d)), d["strength"], d["model_rank"], d["inter"]])
    y = d["exec_match"].values.astype(float)
    beta, *_ = np.linalg.lstsq(X, y, rcond=None)
    resid = y - X @ beta
    dof = len(y) - X.shape[1]
    sigma2 = (resid @ resid) / dof
    cov = sigma2 * np.linalg.inv(X.T @ X)
    se = np.sqrt(np.diag(cov))
    t = beta / se
    p = 2 * (1 - stats.t.cdf(np.abs(t), dof))
    names = ["intercept", "strength", "model_rank", "strength_x_model"]
    return pd.DataFrame({"dataset": dataset, "term": names,
                         "coef": beta.round(4), "se": se.round(4),
                         "t": t.round(3), "p": p.round(4)})

ols_all = []
for ds in sorted(res["dataset"].unique()):
    o = ols_interaction(res, ds)
    if o is not None:
        ols_all.append(o)
ols_tbl = pd.concat(ols_all, ignore_index=True) if ols_all else pd.DataFrame()
display(ols_tbl)
save_table(ols_tbl, "09_pruning_ols")
print("Expect: 'strength' coef negative & significant; "
      "'strength_x_model' NOT significant (p > 0.05).")


# %% [markdown]
# ## Cell 8 - (4) Bootstrap 95% CI for headline EX numbers
# Per (dataset, condition, model): bootstrap the mean EX over questions.

# %%
def bootstrap_ex(df, dataset, condition, model_tier):
    sub = df[(df["dataset"] == dataset) & (df["condition"] == condition) &
             (df["model_tier"] == model_tier)].dropna(subset=["exec_match"])
    y = sub["exec_match"].values.astype(float)
    if len(y) == 0:
        return None
    boot = [np.mean(np.random.choice(y, size=len(y), replace=True))
            for _ in range(N_BOOT)]
    lo, hi = np.percentile(boot, [2.5, 97.5])
    return {"dataset": dataset, "condition": condition, "model_tier": model_tier,
            "n": len(y), "EX": round(y.mean(), 4),
            "CI_low": round(lo, 4), "CI_high": round(hi, 4)}

HEADLINE_CONDS = ["C1", "C2", "C3", "C4", "C5_L1", "C5_L2", "C5_L3"]
rows = []
for ds in sorted(res["dataset"].unique()):
    for cond in HEADLINE_CONDS:
        for mt in MODEL_TIER_ORDER:
            r = bootstrap_ex(res, ds, cond, mt)
            if r:
                rows.append(r)
ex_ci = pd.DataFrame(rows)
display(ex_ci)
save_table(ex_ci, "09_bootstrap_ex_ci")


# %% [markdown]
# ## Cell 9 - Forest plot: EX with 95% bootstrap CI (grayscale)
# One panel per dataset; conditions on the y-axis, EX with CI whiskers, one row
# group per model tier. Lets reviewers see overlap between C1 and C4 at a glance.

# %%
def forest_plot(df_ci, dataset):
    d = df_ci[df_ci["dataset"] == dataset]
    if not len(d):
        return
    conds = [c for c in HEADLINE_CONDS if c in d["condition"].unique()]
    tiers = [t for t in MODEL_TIER_ORDER if t in d["model_tier"].unique()]
    fig, ax = plt.subplots(figsize=(6.2, 0.5 * len(conds) * len(tiers) + 1.2))
    grays = gray_palette(len(tiers))
    ylab, yt = [], []
    y = 0
    for cond in conds:
        for i, mt in enumerate(tiers):
            r = d[(d["condition"] == cond) & (d["model_tier"] == mt)]
            if not len(r):
                continue
            r = r.iloc[0]
            ax.errorbar(r["EX"], y,
                        xerr=[[r["EX"] - r["CI_low"]], [r["CI_high"] - r["EX"]]],
                        fmt=MARKERS[i % len(MARKERS)], color=grays[i],
                        markeredgecolor="black", markersize=6, capsize=3,
                        elinewidth=1.0)
            ylab.append(f"{cond} · {MODEL_TIER_LABELS.get(mt, mt)}")
            yt.append(y)
            y += 1
        y += 0.4  # gap between condition groups
    # reference line at C1/STRONG EX if present
    ref = d[(d["condition"] == "C1") & (d["model_tier"] == "STRONG")]
    if len(ref):
        ax.axvline(ref.iloc[0]["EX"], color="black", linewidth=0.6, linestyle="--")
    ax.set_yticks(yt)
    ax.set_yticklabels(ylab, fontsize=7)
    ax.set_xlabel("Execution Accuracy (95% bootstrap CI)")
    ax.set_title(DATASETS.get(dataset, dataset))
    ax.invert_yaxis()
    save_figure(fig, f"09_ex_ci_forest_{dataset}")
    plt.show()

for ds in sorted(res["dataset"].unique()):
    forest_plot(ex_ci, ds)


# %% [markdown]
# ## Cell 10 - Plain-language summary of the confirmatory tests
# Prints a compact verdict per dataset for the two headline claims.

# %%
print("=== Confirmatory summary ===")
for ds in sorted(res["dataset"].unique()):
    # parity (STRONG)
    pr = parity[(parity["dataset"] == ds) & (parity["model_tier"] == "STRONG")]
    ts = tost[(tost["dataset"] == ds) & (tost["model_tier"] == "STRONG")]
    tk = tok_tests[(tok_tests["dataset"] == ds) & (tok_tests["condition"] == "C4")]
    line = f"[{ds}] "
    if len(pr):
        line += (f"C4 vs C1 (STRONG): dEX={pr.iloc[0]['delta_EX']:+.3f}, "
                 f"McNemar p={pr.iloc[0]['mcnemar_p']:.3f}")
    if len(ts):
        line += (f"; TOST equivalent={ts.iloc[0]['equivalent']} "
                 f"(margin {EQUIV_MARGIN})")
    if len(tk):
        line += (f"; C4 tokens saved={tk.iloc[0]['mean_pct_saved']:.1f}% "
                 f"[{tk.iloc[0]['pct_saved_CI_low']:.1f},"
                 f"{tk.iloc[0]['pct_saved_CI_high']:.1f}]")
    print(line)
```
